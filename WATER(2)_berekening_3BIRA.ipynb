Dit script mag gebruikt worden voor de berekening van opvang-, infiltratie-, en buffercapaciteit van hemelwater in het kader van de DZH opdracht voor Architectuurontwerp 3A.<br>
Er wordt verwacht dat (1) het eindverbruik en de eindcapaciteit voor het hergebruik van regenwater, en (2) het infiltatie- en buffervolume en de infiltratieoppervlakte worden berekend (zie opdracht).<br>
Dit script loodst je door die berekeningen en geeft aan wanneer input van het ontwerpteam is vereist.

**(2) De infiltratie- en buffercapaciteit**

Stap 1. Bepaal het afwaterend oppervlak. (**Deze cel moet aangepast worden**)

```python
verharding_en_terrassen_site = 60
afwaterend_dakoppervlak = 780
groendaken = 60
totaal_aantal_wooneenheden_aangesloten_op_hemelwaterputten = 25

gereduceerd_afwaterend_dakoppervlak = afwaterend_dakoppervlak - 30 * totaal_aantal_wooneenheden_aangesloten_op_hemelwaterputten if afwaterend_dakoppervlak - 30 * totaal_aantal_wooneenheden_aangesloten_op_hemelwaterputten > 0 else 0
afwaterend_oppervlak = verharding_en_terrassen_site + gereduceerd_afwaterend_dakoppervlak + 0.5 * groendaken
print(f"Afwaterend oppervlak = {afwaterend_oppervlak}m²\n")
```

Stap 2. Bepaal de infiltratie- en buffercapaciteit

```python
infiltratieoppervlakte = 0.08 * afwaterend_oppervlak
infiltratievolume = 33 * afwaterend_oppervlak
buffervolume = 43 * afwaterend_oppervlak
print(f"Minimaal infiltratieoppervlakte = {infiltratieoppervlakte}m²\nMinimaal infiltratievolume = {int(infiltratievolume)} l \nMinimaal buffervolume = {int(buffervolume)} l")
```
