# UHI berekening 3BIRA

Dit script mag gebruikt worden voor de berekening van het UHI effect in het kader van de DZH opdracht voor Architectuurontwerp 3A.  <br>
Er wordt verwacht dat het lokale verkoelend effect en het lokale UHI effect van zowel de gehele site als drie locaties op die site (met straal 30m) worden berekend (zie opdracht).<br>
Dit script loodst je door die berekeningen en geeft aan wanneer input van het ontwerpteam is vereist.

Als eerste stap leggen we de variabelen 'maximaal gemiddeld UHI effect' (in een straal van 10km rond de site) en 'potentieel UHI effect' (in een straal van 1km rond de site) vast.

```python
Maximum_UHI = 3.88
Potential_UHI = 3.10
```

Als tweede stap definieren we 2 functies. <br>
De eerste functie berekent de reductiefactor op basis van de reductiefactoren van verschillende terreintypes. Met deze reductiefactor kan het verkoelend effect en daarna het locaal UHI effect berekend worden.
De tweede functie controleert of de ingegeven terreinaandelen samen 100% vormen, zodat er geen foutieve resultaten worden gegenereerd door een typfout of vergissing.

```python
def bereken_fr_reduction(terrein_percentages_site, reductiefactoren):
    fr = 0
    for terrein, percentage in terrein_percentages_site.items():
        factor = reductiefactoren.get(terrein, 0)
        fr += (percentage / 100) * factor
    return fr
    
def controleer_terreinpercentages(terrein_dict, naam):
    totaal = sum(terrein_dict.values())
    if round(totaal, 2) != 100:
        raise ValueError(
            f"Terreintotaal voor {naam} is {round(totaal, 2)}%, geen 100%!"
        )
    else:
        return True
```

Hier volgen de reductiefactoren voor verschillende types terrein. In deze cel hoef je niets aan te passen.

```python
reductiefactoren = {
    "Trees": 0.5,
    "Shrubs and bushes": 0.3,
    "Low vegetation": 0.2,
    "Green roof": 0.2,
    "Build-up area": 0,
    "(Semi-)natural vegetation": 0.2,
    "Inland water": 0.3,
    "Sea": 1,
    "Agricultural land": 0.225,  # gemiddelde van 0.15-0.30
    "Bare soil": 0,
}
```

Als derde stap moet de diversiteit aan terrein types worden ingevoerd. <br>
Deze cel bevat de percentages van het aandeel van elk van de types terrein. **Deze cel moet aangepast worden** naar de verhoudingen van de volledige site.<br>
Een aandeel mag ook nul zijn (vb. sea in het voorbeeld). Indien je kommagetallen invoegt, doe dit dan met een punt ipv een komma (vb. vegetation en inland water in het voorbeeld).

```python
terrein_percentages_site = {
    "Trees": 25,
    "Shrubs and bushes": 10,
    "Low vegetation": 5,
    "Green roof": 5,
    "Build-up area": 40,
    "(Semi-)natural vegetation": 9.5,
    "Inland water": 5.5,
    "Sea": 0,
    "Agricultural land": 0,
    "Bare soil": 0,
}
controle1 = controleer_terreinpercentages(terrein_percentages_site, "site")
```

Dezelfde oefening moet nu gedaan worden voor 3 plekken of iteraties van lokale plaatsen met straal 30m op de site.<br>
LOKALE PLAATS A (**Deze cel moet aangepast worden**)

```python
terrein_percentages_30m_A = {
    "Trees": 35,
    "Shrubs and bushes": 30,
    "Low vegetation": 15,
    "Green roof": 0,
    "Build-up area": 0,
    "(Semi-)natural vegetation": 10,
    "Inland water": 10,
    "Sea": 0,
    "Agricultural land": 0,
    "Bare soil": 0,
}
controle2 = controleer_terreinpercentages(terrein_percentages_30m_A, "30 meter zone A")
```

LOKALE PLAATS B (**Deze cel moet aangepast worden**).

```python
terrein_percentages_30m_B = {
    "Trees": 25,
    "Shrubs and bushes": 10,
    "Low vegetation": 5,
    "Green roof": 5,
    "Build-up area": 40,
    "(Semi-)natural vegetation": 9.5,
    "Inland water": 5.5,
    "Sea": 0,
    "Agricultural land": 0,
    "Bare soil": 0
}
controle3 = controleer_terreinpercentages(terrein_percentages_30m_B, "30 meter zone B")
```

LOKALE PLAATS C (**Deze cel moet aangepast worden**).

```python
terrein_percentages_30m_C = {
    "Trees": 0,
    "Shrubs and bushes": 0,
    "Low vegetation": 0,
    "Green roof": 0,
    "Build-up area": 100,
    "(Semi-)natural vegetation": 0,
    "Inland water": 0,
    "Sea": 0,
    "Agricultural land": 0,
    "Bare soil": 0
}
controle4 = controleer_terreinpercentages(terrein_percentages_30m_C, "30 meter zone C")
```

In de vierde en laatste stap wordt via de functie uit stap 1 de reductiefactor berekend voor de site en elk van de lokale plaatsen (A, B en C). <br>
Met die reductiefactoren kan het "In situ cooling effect vegetation and water"

```python
fr_Reduction_site = bereken_fr_reduction(terrein_percentages_site, reductiefactoren)
In_situ_cooling_effect_vegetation_and_water = Potential_UHI * fr_Reduction_site
fr_Reduction_30m_A = bereken_fr_reduction(terrein_percentages_30m_A, reductiefactoren)
fr_Reduction_30m_B = bereken_fr_reduction(terrein_percentages_30m_B, reductiefactoren)
fr_Reduction_30m_C = bereken_fr_reduction(terrein_percentages_30m_C, reductiefactoren)
Actual_local_UHI_A = Potential_UHI * (1-fr_Reduction_30m_A)
Actual_local_UHI_B = Potential_UHI * (1-fr_Reduction_30m_B)
Actual_local_UHI_C = Potential_UHI * (1-fr_Reduction_30m_C)
Local_cooling_effect_of_vegetation_and_water_A = Maximum_UHI - Actual_local_UHI_A
Local_cooling_effect_of_vegetation_and_water_B = Maximum_UHI - Actual_local_UHI_B
Local_cooling_effect_of_vegetation_and_water_C = Maximum_UHI - Actual_local_UHI_C
```

In de volgende dictionary 'res' worden de resultaten opgenomen

```python
res = {
        "fr Reduction (site)": round(fr_Reduction_site, 2),
        "In situ cooling effect vegetation and water": round(In_situ_cooling_effect_vegetation_and_water, 2),
        "fr Reduction (locatie A)": round(fr_Reduction_30m_A, 2),
        "fr Reduction (locatie B)": round(fr_Reduction_30m_B, 2),
        "fr Reduction (locatie C)": round(fr_Reduction_30m_C, 2),
        "Actual UHI (locatie A)": round(Actual_local_UHI_A, 2),
        "Actual UHI (locatie B)": round(Actual_local_UHI_B, 2),
        "Actual UHI (locatie C)": round(Actual_local_UHI_C, 2),
        "Local cooling effect of vegetation and water (locatie A)": round(Local_cooling_effect_of_vegetation_and_water_A, 2),
        "Local cooling effect of vegetation and water (locatie B)": round(Local_cooling_effect_of_vegetation_and_water_B, 2),
        "Local cooling effect of vegetation and water (locatie C)": round(Local_cooling_effect_of_vegetation_and_water_C, 2)
    }
```

De functies print_site_overzicht en print_lokale_overizcht zorgen voor de output van de resultaten van de volledige site en de drie lokale plekken, resp.

```python
def print_site_overzicht(naam, terreinpercentages, fr, cooling):
    print("\n" + "="*80)
    print(naam)
    print("="*80)
    print(f"{'Terrain type':<30} {'Share (%)':>10} {'Reduction factor':>18} {'Contribution':>15}")
    print("-"*80)
    for terrein, aandeel in terreinpercentages.items():
        factor = reductiefactoren[terrein]
        bijdrage = aandeel / 100 * factor
        if aandeel > 0:
            print(
                f"{terrein:<30} "
                f"{aandeel:>10.1f} "
                f"{factor:>18.3f} "
                f"{bijdrage:>15.3f}")
    print("-"*80)
    print(f"{'Reduction factor (-)':<50} {fr:.2f}")
    print(f"{'In situ cooling effect vegetation and water (°C)':<50} {cooling:.2f}°C")


def print_lokale_overzicht(naam, terreinpercentages, fr, uhi, cooling):
    print("\n" + "="*80)
    print(naam)
    print("="*80)
    print(f"{'Terrain type':<30} {'Share (%)':>10} {'Reduction factor':>18} {'Contribution':>15}")
    print("-"*80)
    for terrein, aandeel in terreinpercentages.items():
        factor = reductiefactoren[terrein]
        bijdrage = aandeel / 100 * factor
        if aandeel > 0:
            print(
                f"{terrein:<30} "
                f"{aandeel:>10.1f} "
                f"{factor:>18.3f} "
                f"{bijdrage:>15.3f}")
    print("-"*80)
    print(f"{'Reduction factor (-)':<50} {fr:.2f}")
    print(f"{'Actual local UHI effect (°C)':<50} {uhi:.2f}°C")
    print(f"{'Local cooling effect vegetation and water (°C)':<50} {cooling:.2f}°C")
```

Indien voldaan is aan alle controles (het totaal van terreinaandelen op de site en de drie lokale plekken is telkens gelijk aan 100%), zal de output geprint worden.<br>
Deze print mag rechtstreeks worden gebruikt als bijlage voor de UHI opdracht.<br>
De resultaten worden gebruikt en geïnterpreteerd in de conceptnota('s).

```python
if controle1 and controle2 and controle3 and controle4:
    print_site_overzicht("SITE", terrein_percentages_site, fr_Reduction_site, In_situ_cooling_effect_vegetation_and_water)
    print_lokale_overzicht("LOCAL PLACE A (30 m radius)", terrein_percentages_30m_A, fr_Reduction_30m_A, Actual_local_UHI_A, Local_cooling_effect_of_vegetation_and_water_A)
    print_lokale_overzicht("LOCAL PLACE B (30 m radius)",terrein_percentages_30m_B,fr_Reduction_30m_B,Actual_local_UHI_B,Local_cooling_effect_of_vegetation_and_water_B)
    print_lokale_overzicht("LOCAL PLACE C (30 m radius)", terrein_percentages_30m_C, fr_Reduction_30m_C, Actual_local_UHI_C, Local_cooling_effect_of_vegetation_and_water_C)
```
